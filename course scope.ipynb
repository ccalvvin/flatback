```python
import requests
import io
import urllib3
from pypdf import PdfReader
# Suppress the SSL verification warning
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
def get_course_pdf_text(course_code):
    """
    Fetches and extracts text from the SUSS course PDF based on the provided code.
    """
    # Dynamically build the URL using the course_code argument
    url = f"https://sims1.suss.edu.sg/eservice/public/viewcourse/viewcourse.aspx?crsecd={course_code}&viewtype=pdf&isft=0"
    
    try:
        print(f"Fetching data for: {course_code}...")
        response = requests.get(url, verify=False)
        response.raise_for_status()
        
        pdf_memory_file = io.BytesIO(response.content)
        reader = PdfReader(pdf_memory_file)
        
        full_text = ""
        for i, page in enumerate(reader.pages):
            text = page.extract_text()
            if text:
                full_text += f"\n--- Page {i+1} ---\n" + text
        
        return full_text
    except requests.exceptions.HTTPError as e:
        return f"HTTP error occurred: {e}"
    except Exception as e:
        return f"An error occurred: {e}"
# --- Usage ---
# Simply change this variable to the course code you want
course_to_fetch = "ANL312" 
content = get_course_pdf_text(course_to_fetch)
# Print the result or save to a file
if content:
    print(f"\n--- Output for {course_to_fetch} ---")
    print(content)
```
